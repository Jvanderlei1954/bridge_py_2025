```python
# List of products, their prices, and the quantities sold
frutas = ["Abacate.....", "Abacaxi.....", "Banana......", "Caqui.......", "Kiwi........", 
          "Laranja_Lima", "Laranja_Pera", "Limão.......", "Macã........", "Mamão.......", "Melão.......", 
          "Ponkam......", "Pêra........", "Uva........."]

precos = [10.00, 12.00, 15.00, 15.00, 12.00, 6.00, 12.00, 2.50, 9.00, 15.00, 30.00, 5.00, 10.00, 25.00]

qtde_vendida = [150, 200, 100, 50, 60, 180, 90, 140, 80, 120, 40, 250, 60, 70]

custo_total = float(0)
ind_custo = 0
global ind_custo

for calc in range(len(precos)):
    calc_total = precos[calc] * qtde_vendida[calc]
    custo_total += calc_total
    ind_custo = custo_total    
 
def calculate_receita(precos, qtde_vendida):
    receita = []
    global custo_total
    
    for i in range(len(precos)):
        valor = precos[i] * qtde_vendida[i]
        perc = (precos[i] / ind_custo) * 100
        receita.append((frutas[i], valor))
        print(f"{frutas[i]} Valor unit. = {precos[i]:.2f} ==> Lote da fruta = R$ {valor:.2f}\
        indice sobre valor estoque/frutas = {perc:.2f}%")
    return receita

def formatted_output(receitas):
    rev_tot = 0
    perc_rev = 0
    for fruta, rev in sorted(receitas):
        rev_tot += rev
        perc_rev = (rev/rev_tot) * 100
    print("\n" + "=" * 60)
    print(f"Custo do estoque/frutas = R$ {rev_tot:.2f}")
    print("=" * 60)
    
# Chamadas principais
receita_total = calculate_receita(precos, qtde_vendida)
formatted_output(receita_total)
```
