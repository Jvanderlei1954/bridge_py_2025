```python
#############################################################################################
# Objetivo.
# Demonstrar como converter diferentes representações de preços (string e inteiro)
# para o tipo float e calcular a proporção entre eles, incluindo a diferença percentual.
#############################################################################################
```

```python
price1 = "3.99" # 10
price2 = 12 # 10

# Convert prices to float
price1_converted = float(price1)
price2_converted = float(price2)

print(f"Price #1 is ${price1_converted:.2f} and is of type {type(price1_converted)}")
print(f"Price #2 is ${price2_converted:.2f} and is of type {type(price2_converted)}")
print(f"Média entre preço1 e preço2 is ${(price2_converted/2)} and is of type {type(price2_converted)}")
#print(f"Proporção entre a diferença de preço1 e preço2 is \
#{(price1_converted/price2_converted):.2f}% and is of type {type(price2_converted)}")
```

```python
price1 = "3.99" 
price2 = 12

# Convert prices to float
price1_converted = float(price1)
price2_converted = float(price2)

print(f"Valor de preço1 é {price1_converted} e de preço2 é {price2_converted}")
prop_preco = price1_converted/price2_converted
percent = (prop_preco * 100)
print(f"Portanto: os preços consiste em uma diferênça de {prop_preco} em percentual igual a {percent}% de preço2")
```
