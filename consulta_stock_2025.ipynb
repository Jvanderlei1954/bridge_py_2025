```python
seasonal = True
on_sale = False
selling_well = False
current_stock = 150
high_stock_threshold = 100

# Calculating reorder point and stock level.
reorder_point = (current_stock -1) - (high_stock_threshold)
percent_stock_baixo = (reorder_point/current_stock)*100
print("Estoque corrente =", current_stock)
print("Limite de estoque =",high_stock_threshold)
print("ponto de pedido =",reorder_point)
print("Percentual do ponto de pedido", f"{percent_stock_baixo:.2f}%")

# Answer if stck can be discounted?
overstock_risk    = seasonal and current_stock > high_stock_threshold
discount_eligible = not selling_well and not on_sale
make_discount     = overstock_risk or discount_eligible
print("Should the item be discounted?", make_discount)

```
