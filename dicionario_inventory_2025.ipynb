```python
# Inventory dictionary with stock, price, and discount price
import math
inventory = {
    "Bread": [42, 1.20, 0.99],  # "Item": [current stock, regular price, discounted price]
    "Eggs": [225, 2.12, 1.99],  # Eggs should be sold at a discount
    "Apples": [9, 1.50, 1.35]   # Apples need to be restocked
}

print("=================================================================================")

for item in inventory:
    current_stock, regular_price, discounted_price = inventory[item]
    bread_val = inventory["Bread"][0] * inventory["Bread"][1]
    eggs_val = inventory["Eggs"][0] * inventory["Eggs"][1]
    apples_val = inventory["Apples"][0] * inventory["Apples"][1]
    
    if current_stock < 30:
        print(f"{item} com {current_stock} unidades estocadas - Próximo de seu reabastecimento.")
    elif current_stock > 100:
        print(f"{item} deve ser vendido pelo preço com desconto de ${discounted_price} ==> Preço normal = ${regular_price}")
    else:
        print(f"{item} deve ser vendido pelo preço normal ${regular_price}.")

# Receita esperada do estoque
print("=================================================================================")
print(f"Receita do estoque de pães a receber: ${bread_val:.2f}")
print(f"Receita do estoque de ovos a receber: ${eggs_val:.2f}")
print(f"Receita do estoque de maçãs a receber: ${apples_val:.2f}")
print("=================================================================================")

total_geral = (bread_val + eggs_val + apples_val)
rec_bread = (bread_val/total_geral * 100)
rec_eggs = (eggs_val/total_geral * 100)
rec_apples = (apples_val/total_geral * 100)

print(f"Participação da receita dos pães sobre o total a receber: {rec_bread:.2f}%")
print(f"Participação da receita dos ovos sobre o total a receber: {rec_eggs:.2f}%")
print(f"Participação da receita das maçãs sobre o total a receber: {rec_apples:.2f}%")
print(f"Participação geral: ${total_geral:.2f} que significa 100% da receita a receber")
print("=================================================================================")
```
