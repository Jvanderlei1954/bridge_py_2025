```python
daily_promotions = ["Milk", "Eggs", "Bread", "Apples", "Oranges", "Bean",
                    "Rice", "Potato", "Fish", "Meet", "Banana", "pineapple"]

custo = [2.55, 12.60, 1.82, 9.00, 3.55, 12.00, 27.70, 8.50, 25.00, 75.40, 15.00, 12.00]

month_yers = ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun", "Jul", "Ago", "Set", "Out", "Nov", "Dez"]

for month in range(12): 
    monthyers = month_yers[month]  # Nome do mês
    promotion = daily_promotions[month]  # Produto em promoção
    desconto = (custo[month] / 5)*4  # Aplica desconto de 20%
    valor = custo[month]
    print(f"{monthyers}: Promotion on {promotion} = ${valor:.2f} com desconto de 20% = ${desconto:.2f}")
```
