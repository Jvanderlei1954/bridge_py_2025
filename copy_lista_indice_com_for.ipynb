```python
#Para garantir que cada elemento da lista prices seja atualizado diretamente , utilizamos range(len(prices))
#para iterar sobre os índices.
```

```python
# List of original prices of grocery items
prices = [1.50, 2.00, 0.75, 3.25]

# Discount factor (10% off each item)
discount_factor = 0.10

# Store original prices before applying discount
price_gross = prices.copy()

# Iterate over the list of prices using range(len())
for cost in range(len(prices)):
    # Apply the discount by reducing the price
    prices[cost] -= prices[cost] * discount_factor
    print(f"Item => {cost + 1} Preço cheio = ${price_gross[cost]:.2f}: Novo preço = ${prices[cost]:.2f}")

prices = [round(price, 2) 
for price in prices]
print("Updated prices:", prices)
```
