```python
#===========================================================================================================
# Processar dados de produtos a partir de um dicionário onde os preços e as quantidades estão armazenados 
# como strings. O objetivo é calcular as vendas totais para cada produto e gerar estatísticas resumidas.
#===========================================================================================================
```

```python
# Dictionary of products with price and quantity sold as strings
products = {
    "Apple": ["1.20", "50"],   # "Item": [price, quantity sold]
    "Banana": ["0.50", "100"],
    "Cherry": ["2.50", "25"],
    "Mango": ["1.75", "40"]
}
# Substitua float(products) por uma lista vazia
total_sales_list = []
print("=============================")
print("Sold prodiucts")
print("=============================")
# Iterar e processar cada produto - com transformação de string em float e int consecutivamente
for product, values in products.items():
#=====================================================================================================
#===> Exemplo: price of products.Apple in "120" is a string e products.quantity_sold in "50" too is.
#=====================================================================================================
    price = float(values[0])
    quantity_sold = int(values[1])
    total_sales = price * quantity_sold
    #print("Sold prodiucts")
    print("============================")
    print(f"Total sales for {product}: ${total_sales}")
    total_sales_list.append(total_sales)

# Estatísticas finais
total_sum = sum(total_sales_list)
min_sales = min(total_sales_list)
max_sales = max(total_sales_list)

print("  ")
print("===== Estatística finais =====")
print(f"\nTotal sum of all sales: ${total_sum}")
print(f"Minimum sales: ${min_sales}")
print(f"Maximum sales: ${max_sales}")
```
