```python
# A função zip(): combina dois ou mais iteráveis como: (listas) em um único iterável de tuplas, 
# agrupando elementos de cada iterável juntos.

products = ["apple", "banana", "cherry", "guava"]
prices = [0.99, 0.59, 2.99, 20]
stock = [50, 100, 25, 70]

# `zip()` combines the 3 lists into a series of tuples
# `list()` converts the zip object into a list
product_info = list(zip(products, prices, stock))

print("Product information:", product_info)
```
