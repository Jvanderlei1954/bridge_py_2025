```python
# A função sorted(): retorna uma nova lista ordenada a partir de um iterável (como listas, tuplas ou dicionários). 
# Diferente do método sort(), sorted(), não modifica os dados originais e funciona com uma variedade maior de tipos.
```

```python
# Aplicando somente a função sorted()
fruit_prices = {"cherries": 3.99, "apples": 2.99, "bananas": 1.49}

# Sorting the dictionary keys alphabetically
sorted1_prices = sorted(fruit_prices)

print(sorted1_prices)
```

```python
# Aplicando função sorted(), associada com a função items(), assim os nomes dos itens e seus preços são apresentados
fruit_prices = {"cherries": 3.99, "apples": 2.99, "bananas": 1.49}

# Sorting the dictionary keys alphabetically
sorted_prices = dict(sorted(fruit_prices.items()))

print("===============================================================================================")
print(f"# Apresentando somente o nome do produto: {sorted1_prices}")
print("===============================================================================================")
print(f"# Apresentando nome do produto e preço: {sorted_prices}")
print("===============================================================================================")
```
