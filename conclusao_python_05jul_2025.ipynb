```python
# -*- coding: utf-8 -*-

# Lista de produtos, preços e quantidades vendidas
produtos              = ["Pães", "Maçãs", "Laranjas", "Bananas"]
precos_por_item       = [0.50, 1.20, 2.50, 2.00]
quantidades_vendidas  = [250, 200, 100, 50]

def calcular_receita(produtos, precos, quantidades):
    """
    Calcula a receita individual de cada produto e a receita total.
    
    Retorna:
        lista de tuplas (produto, receita_individual)
    """
    receita_por_produto = []
    receita_total = 0

    print("* Inventário de junho de 2025 ".center(75, "*"))
    for nome, preco, qtde in zip(produtos, precos, quantidades):
        receita = preco * qtde
        receita_por_produto.append((nome, receita))
        receita_total += receita
        print(f"{nome:<10} | Qtde vendida: {qtde:>4} | "
              f"Preço unit.: R${preco:>5.2f} | Receita: R${receita:>7.2f}")
    print("*" * 75)
    print(f"Receita total obtida: R${receita_total:.2f}\n")

    return receita_por_produto

# -------------------- Execução principal --------------------
receitas = calcular_receita(produtos, precos_por_item, quantidades_vendidas)

print("          Receita por produto (da maior para a menor):")
#print("=" * 54)
print("         =============================================")
for nome, rec in sorted(receitas, key=lambda x: x[1], reverse=True):
    print(f"                   • {nome:<10} → R${rec:.2f}")
print("         =============================================")
```
