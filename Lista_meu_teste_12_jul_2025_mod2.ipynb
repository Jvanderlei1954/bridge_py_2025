```python
frutas = ["Abacate.....", "Abacaxi.....", "Banana......", "Caqui.......", "Kiwi........", 
          "Laranja_Lima", "Laranja_Pera", "Limão.......", "Macã........", "Mamão.......", 
          "Melão.......", "Ponkam......", "Pêra........", "Uva........."]

precos = [10.00, 12.00, 15.00, 15.00, 12.00, 6.00, 12.00, 2.50, 9.00, 15.00, 30.00, 5.00, 10.00, 25.00]
qtde_vendida = [150, 200, 100, 50, 60, 180, 90, 140, 80, 120, 40, 250, 60, 70]

custo_total = 0
calc_ind = 0.0
rev_tot = 0.0
global rev_tot2

def formatted_output(receitas):
    rev_tot = sum(rev for _, rev in receitas)
    print("\n" + "=" * 60)
    print(f"Custo total do estoque de frutas = R$ {rev_tot:.2f}")
    print("=" * 60)
    #print(rev_tot)

def calculate_receita(precos, qtde_vendida):
    global calc_ind
    global custo_total
    receita = []
    print("\n=== Detalhamento por produto ===\n")

    for i in range(len(precos)):
        calc_total = precos[i] * qtde_vendida[i]
        custo_total += calc_total
        calc_ind = (precos[i] / calc_total) * 100  # Percentual do unitário sobre o lote
        receita.append((frutas[i], calc_total))
        print(f"{frutas[i]} Valor unitário = R$ {precos[i]:.2f} ==> Lote da fruta = R$ {calc_total:.2f}")
        print(f"  --> Percentual do preço unitário sobre o lote: {calc_ind:.2f}%")
        print("-" * 60)

    return receita

# Chamadas principais
receita_total = calculate_receita(precos, qtde_vendida)
formatted_output(receita_total)
```
